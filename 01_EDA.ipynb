Frequency

```python
total_policies = len(freq)
total_exposure = freq["Exposure"].sum()
total_claims = freq["ClaimNb"].sum()

overall_frequency = total_claims / total_exposure

print("Policies:", total_policies)
print("Exposure:", total_exposure)
print("Claims:", total_claims)
print("Claim frequency:", overall_frequency)
```

Severity

```python
total_loss = sev["ClaimAmount"].sum()
average_severity = sev["ClaimAmount"].mean()
median_severity = sev["ClaimAmount"].median()

print("Total Loss:", total_loss)
print("Mean Severity:", average_severity)
print("Median Severity:", median_severity)
```
